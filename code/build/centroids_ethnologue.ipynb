# Centroid lat/lon per Ethnologue polygon

Computes `lat` and `lon` (degrees) for each Ethnologue polygon and exports a CSV for merging into the Stata build. These centroids are intended for Conley (1999) spatial HAC standard errors.

Centroids are computed in EPSG:6933 (equal-area) and then reprojected to EPSG:4326 to extract lon/lat — this avoids the bias you get when taking centroids directly in a geographic CRS on large polygons.

```python
from pathlib import Path
import geopandas as gpd
```

```python
base_path = Path("C:/Users/juami/Dropbox/RAships/2-Folklore-Nathan-Project/EA-Maps-Nathan-project/Measures_work")

poscol_path = base_path / "data" / "raw" / "ethnologue" / "ancestral_characteristics_database_language_level" / "Ethnologue_16_shapefile" / "langa_no_overlap_biggest_clean.shp"

out_path = base_path / "maps" / "raw" / "Geography"
out_path.mkdir(parents=True, exist_ok=True)
```

```python
poscol = gpd.read_file(poscol_path)[["ID", "geometry"]]

# Centroid in equal-area CRS, then reproject to 4326 to extract lon/lat
centroids = poscol.to_crs(6933).geometry.centroid.to_crs(4326)

poscol["lat"] = centroids.y
poscol["lon"] = centroids.x

poscol[["ID", "lat", "lon"]].describe()
```

```python
out_csv = out_path / "ethnologue_centroids.csv"
poscol[["ID", "lat", "lon"]].to_csv(out_csv, index=False)
print("Exported", out_csv)
```
