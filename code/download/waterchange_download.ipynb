# Downloading JRC Global Surface Water layers (for waterchange_ethnologue.ipynb)

Exports go to Google Drive via Earth Engine. Monitor progress at the 
[GEE Task Manager](https://code.earthengine.google.com/tasks).

```python
import os
import ee
import geemap 
import pandas as pd

ee.Authenticate()
ee.Initialize()
```

```python
# Define region to view (Peru/Brazil Amazon)
world_bbox = ee.Geometry.BBox(-180, -85, 180, 85)

# Load the water change band
permwater = ee.Image("JRC/GSW1_4/GlobalSurfaceWater").select("transition").eq(1)
losswater = ee.Image("JRC/GSW1_4/GlobalSurfaceWater").select("transition").eq(3)
permseasonal = ee.Image("JRC/GSW1_4/GlobalSurfaceWater").select("transition").eq(8)

changewaterabs = ee.Image("JRC/GSW1_4/GlobalSurfaceWater").select("change_abs")
changewater = ee.Image("JRC/GSW1_4/GlobalSurfaceWater").select("change_norm")
```

```python
ALREADY DOWNLOADED — skip
task = ee.batch.Export.image.toDrive(
    image=permwater,
    description='perm_water_100m_30m',
    folder='GEE_exports',
    fileNamePrefix='perm_water_100m_30m',
    region=world_bbox,
    scale=100,
    maxPixels=1e13
)
task.start()
```

```python
ALREADY DOWNLOADED — skip
task = ee.batch.Export.image.toDrive(
    image=losswater,
    description='loss_water_100m_30m',
    folder='GEE_exports',
    fileNamePrefix='loss_water_100m_30m',
    region=world_bbox,
    scale=100,
    maxPixels=1e13
)
task.start()
```

```python
# Export to Google Drive — transition == 8 (permanent in 1984-1999, became seasonal in 2000-2021)
task = ee.batch.Export.image.toDrive(
    image=permseasonal,
    description='permseasonal_100m_30m',
    folder='GEE_exports',
    fileNamePrefix='permseasonal_100m_30m',
    region=world_bbox,
    scale=100,
    maxPixels=1e13
)

task.start()
```

```python
ALREADY DOWNLOADED — skip
task = ee.batch.Export.image.toDrive(
    image=changewater,
    description='change_water_100m_30m',
    folder='GEE_exports',
    fileNamePrefix='change_water_100m_30m',
    region=world_bbox,
    scale=100,
    maxPixels=1e13
)
task.start()
```

```python
# NEEDS RUNNING — exports the JRC `change_abs` band (absolute change), in
# parallel to the `change_norm` export above. The two bands give different
# values per polygon; the ethnologue notebook computes both as `changewater`
# (norm) and `changewater_abs` (abs) so they can be compared.
# After GEE finishes, download the .tif tiles from Drive and place them under
#   maps/raw/Water_surface/change_water_abs/
task = ee.batch.Export.image.toDrive(
    image=changewaterabs,
    description='change_water_abs_100m_30m',
    folder='GEE_exports',
    fileNamePrefix='change_water_abs_100m_30m',
    region=world_bbox,
    scale=100,
    maxPixels=1e13
)
task.start()
```
